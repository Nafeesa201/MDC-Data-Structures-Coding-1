1. Numbers between 2000 and 3200 divisible by 7 and not a multiple of 3

```python
result=[]
for no in range(2000, 3201):
  if no%7==0 and no%3!=0:
    result.append(str(no))
print(",".join(result))
```

2. Convert a List of Characters into a String


```python
char_list=["N","A","F","E","E","S","A"]
result_str=""
for char in char_list:
  result_str += char
print(result_str.capitalize())
```

3. Generate a Dictionary of Squares (i, i x i)

```python
no=10
dictionary={}
for i in range(1,no+1):
  dictionary[i]=i*i
print(dictionary)
```
